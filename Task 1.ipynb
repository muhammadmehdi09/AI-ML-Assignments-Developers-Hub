# Exploring and Visualizing a Simple Dataset

## 1. Import

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load Dataset

```python
data = sns.load_dataset('iris')
print("Columns: ", data.columns.to_list())
print("Shape: ", data.shape)
```

## 3. Data Exploration

```python
print("Info:")
data.info
```

```python
print("Description:")
data.describe()
```

```python
print("First five rows:")
data.head()
```

## 4. Visualization

```python
sns.scatterplot(
    data=data,
    x='sepal_length',
    y='petal_length',
    hue=data.species,
    palette="bright"
)
plt.title('Scatter Plot: Sepal Length vs Petal Length')
plt.show()
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
plt.style.use("dark_background")
sns.set_palette("deep")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for i, feature in enumerate(features):
    sns.boxplot(
        data=data,
        x='species',
        y=feature,
        ax=axes[i // 2][i % 2],
        hue="species"
    )
    axes[i // 2][i % 2].set_title(f'Box Plot: {feature}')
    axes[i // 2][i % 2].set_xlabel('Species')
    axes[i // 2][i % 2].set_ylabel(feature)

plt.suptitle('Box Plots', fontsize=20)
plt.tight_layout()
plt.show()
```
