# General Health Query Chatbot

## 1. Import

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

## 2. Load API Key

```python
load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

## 3. Chat Function

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful medical assistant. Keep answers short. Do not diagnose. Do not give answers unrelated to medical say I dont know"
    }
]

def ask_bot(user_input):
    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=messages,
        max_tokens=120,
        temperature=0.5
    )

    reply = response.choices[0].message.content
    messages.append({"role": "assistant", "content": reply})
    return reply
```

## 4. Main Loop

```python
print("Health Chatbot Ready (type 'exit' to quit)\n")

danger_keywords = [
    "chest pain",
    "can't breathe",
    "severe bleeding",
    "fainting",
    "stroke",
    "heart attack"
]

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("\nBot: Take care.")
        break

    if any(i in user_input.lower() for i in danger_keywords):
        print("\nBot: This may be serious. Please seek immediate medical attention.\n")
        
    else:
        reply = ask_bot(user_input)
        print("\nBot:", reply, "\n")
```
